```python
import pandas as pd

# 1. 인터넷에서 SECOM 센서 데이터와 합격/불합격 라벨 불러오기
data_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/secom/secom.data"
label_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/secom/secom_labels.data"

X = pd.read_csv(data_url, sep=" ", header=None)
y = pd.read_csv(label_url, sep=" ", header=None, names=["status", "timestamp"])

# 2. 계획서 Step 1: 전체 데이터 크기(행, 열)와 정상(-1)/불량(1) 개수 확인하기
print("센서 데이터 크기 (샘플 수, 센서 개수):", X.shape)
print("\n[정상(-1) vs 불량(1) 개수]")
print(y["status"].value_counts())
```
