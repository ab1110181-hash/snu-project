```python
import kagglehub
import pandas as pd
import os

# 1. 캐글에서 '반도체 웨이퍼 불량 분류' 데이터셋 즉시 다운로드
path = kagglehub.dataset_download("meruvakodandasuraj/semiconductor-wafer-defect-classification-dataset")
print("다운로드 완료된 경로:", path)

# 2. 폴더 안에 들어있는 CSV 파일 이름 찾아서 Pandas 표(DataFrame)로 읽기
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
file_path = os.path.join(path, csv_files[0])
df = pd.read_csv(file_path)

# 3. 데이터 크기와 실제 센서 이름(컬럼명) 상위 5줄 확인하기
print("\n데이터 크기 (웨이퍼 수, 변수 개수):", df.shape)
display(df.head())
```

```python
# 1. 어떤 센서와 공정 정보들이 있는지 전체 컬럼 이름 확인
print("[전체 변수 목록]")
print(df.columns.tolist())

# 2. 데이터 결측치(비어있는 값)가 있는지 확인
print("\n[변수별 결측치 개수]")
print(df.isnull().sum())

# 3. 정상(0) vs 불량(1) 웨이퍼 개수 및 비율 확인 (마지막 열이 불량 라벨)
target_col = df.columns[-1]
print(f"\n[{target_col} 분포 (정상 vs 불량)]")
print(df[target_col].value_counts())

# 4. 정상 웨이퍼와 불량 웨이퍼의 센서 측정값 평균 비교
print("\n[정상 vs 불량 웨이퍼의 센서 평균값 차이]")
display(df.groupby(target_col).mean(numeric_only=True))
```
